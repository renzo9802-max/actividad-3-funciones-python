## 1. `pow()`

**¿Para qué sirve?**  
Calcula una potencia: eleva una base a un exponente.

**Sintaxis:**
```python
pow(base, exponente)
```

**Ejemplo práctico:**  
pow(3, 2) equivale a 3 elevado a la 2, cuyo resultado es 9. También se puede usar el operador **.

````python
filas = 3
plantas_por_fila = 3
print("Total de posiciones en una matriz cuadrada:", pow(plantas_por_fila, 2))
````

## 2. `divmod()`

**¿Para qué sirve?**  
Devuelve al mismo tiempo el cociente entero y el residuo de una división.

**Sintaxis:**
```python
divmod(dividendo, divisor)
```

**Ejemplo práctico:**  
El resultado es una pareja (cociente, residuo). Facilita repartir unidades en grupos del mismo tamaño.

````python
plantas = 125
plantas_por_bandeja = 12
bandejas_completas, sobrantes = divmod(plantas, plantas_por_bandeja)
print("Bandejas completas:", bandejas_completas)
print("Plantas sobrantes:", sobrantes)
````

## 3. `sorted()`

**¿Para qué sirve?**  
Devuelve una nueva lista con los elementos ordenados. Por defecto los organiza de menor a mayor.

**Sintaxis:**
```python
sorted(iterable, reverse=False)
```

**Ejemplo práctico:**  
No modifica la lista original. El argumento reverse=True cambia el orden a descendente.

````python
precios = [1300, 1100, 1500, 1200]
print("Precios ordenados:", sorted(precios))
print("De mayor a menor:", sorted(precios, reverse=True))
````

## 4. `reversed()`

**¿Para qué sirve?**  
Permite recorrer los elementos de una secuencia en orden inverso.

**Sintaxis:**
```python
reversed(secuencia)
```

**Ejemplo práctico:**  
reversed() devuelve un iterador; por eso se usa list() para mostrar todos los elementos en este ejemplo.

````python
dias = ["lunes", "martes", "miércoles", "jueves"]
print(list(reversed(dias)))
````

## 5. `all()`

**¿Para qué sirve?**  
Devuelve True si todos los elementos de una colección son verdaderos. Si al menos uno es falso, devuelve False.

**Sintaxis:**
```python
all(iterable)
```

**Ejemplo práctico:**  
Es útil para comprobar si se cumplen todos los requisitos de una revisión.

````python
lotes_revisados = [True, True, True]
print("¿Todos los lotes fueron revisados?", all(lotes_revisados))
lotes_revisados = [True, False, True]
print("¿Todos los lotes fueron revisados?", all(lotes_revisados))
````

## 6. `any()`

**¿Para qué sirve?**  
Devuelve True si al menos un elemento de la colección es verdadero. Devuelve False si ninguno lo es.

**Sintaxis:**
```python
any(iterable)
```

**Ejemplo práctico:**  
A diferencia de all(), no exige que todos los elementos sean verdaderos; basta con uno.

````python
alertas_de_inventario = [False, True, False]
print("¿Existe alguna alerta?", any(alertas_de_inventario))
````

## 7. `isinstance()`

**¿Para qué sirve?**  
Comprueba si un valor pertenece a un tipo de dato o a una clase determinada.

**Sintaxis:**
```python
isinstance(valor, tipo)
```

**Ejemplo práctico:**  
Devuelve un valor lógico. Es útil para validar el tipo de un dato antes de procesarlo.

````python
unidades = 150
precio = 1250.5
print(isinstance(unidades, int))
print(isinstance(precio, float))
print(isinstance(precio, int))
````

## 8. `zip()`

**¿Para qué sirve?**  
Combina elementos de varias colecciones posición por posición.

**Sintaxis:**
```python
zip(iterable1, iterable2, ...)
```

**Ejemplo práctico:**  
El primer elemento de una colección se combina con el primero de la otra. El recorrido termina cuando se agota la colección más corta.

````python
dias = ["lunes", "martes", "miércoles"]
produccion = [120, 135, 128]
for dia, cantidad in zip(dias, produccion):
    print(dia, "-", cantidad, "unidades")
````

## 9. `map()`

**¿Para qué sirve?**  
Aplica una operación o función a cada elemento de una colección.

**Sintaxis:**
```python
map(funcion, iterable)
```

**Ejemplo práctico:**  
En el ejemplo, lambda define una operación pequeña: multiplicar cada cantidad de cajas por 12. list() permite visualizar todos los resultados.

````python
cajas = [5, 8, 10]
plantas_por_caja = 12
plantas = list(map(lambda cantidad: cantidad * plantas_por_caja, cajas))
print(plantas)
````

## 10. `filter()`

**¿Para qué sirve?**  
Selecciona los elementos de una colección que cumplen una condición.

**Sintaxis:**
```python
filter(funcion_de_condicion, iterable)
```

**Ejemplo práctico:**  
La condición kg >= 10 determina qué elementos se conservan. list() convierte el resultado en una lista visible.

````python
produccion_kg = [8, 12, 5, 15, 9]
produccion_alta = list(filter(lambda kg: kg >= 10, produccion_kg))
print("Registros de 10 kg o más:", produccion_alta)
````

## 11. `format()`

**¿Para qué sirve?**  
Convierte un valor a texto aplicando un formato, por ejemplo, para controlar la cantidad de decimales.

**Sintaxis:**
```python
format(valor, especificacion)
```

**Ejemplo práctico:**  
La especificación .2f muestra dos decimales. format() devuelve texto, por lo que el resultado se puede unir a una etiqueta.

````python
precio_promedio = 1250.5678
precio_mostrado = format(precio_promedio, ".2f")
print("Precio promedio: $" + precio_mostrado)
````

## 12. `repr()`

**¿Para qué sirve?**  
Devuelve una representación de texto del objeto, útil para observar con claridad caracteres especiales como saltos de línea.

**Sintaxis:**
```python
repr(objeto)
```

**Ejemplo práctico:**  
A diferencia de print() aplicado directamente al texto, repr() muestra el carácter de salto de línea como \n dentro de la representación.

````python
descripcion = "Lote A\nLechuga"
print(repr(descripcion))
````

## 13. `bin()`

**¿Para qué sirve?**  
Convierte un número entero a una cadena que representa su valor en sistema binario.

**Sintaxis:**
```python
bin(numero_entero)
```

**Ejemplo práctico:**  
El prefijo 0b indica que el resultado está escrito en base 2.

````python
numero = 10
print(bin(numero))
````

## 14. `hex()`

**¿Para qué sirve?**  
Convierte un número entero a una cadena que representa su valor en sistema hexadecimal.

**Sintaxis:**
```python
hex(numero_entero)
```

**Ejemplo práctico:**  
El prefijo 0x indica que el resultado está escrito en base 16.

````python
numero = 255
print(hex(numero))
````

## 15. `ord()`

**¿Para qué sirve?**  
Devuelve el código Unicode correspondiente a un carácter.

**Sintaxis:**
```python
ord(caracter)
```

**Ejemplo práctico:**  
Recibe una cadena de un solo carácter. Es la operación inversa de chr(), que convierte un código Unicode en carácter.

````python
letra = "A"
print("Código Unicode de A:", ord(letra))
````

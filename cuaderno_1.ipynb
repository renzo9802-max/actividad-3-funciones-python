## 1. `print()`

**¿Para qué sirve?**  
Muestra información en la pantalla. Se usa para comunicar resultados, mensajes y valores al usuario.

**Sintaxis:**
```python
print(objeto)
```

**Ejemplo práctico:**  
Permite mostrar uno o varios datos separados por comas. Python coloca un espacio entre ellos y termina la línea.

````python
cultivo = "Lechuga crespa"
unidades = 120
print("Cultivo:", cultivo)
print("Unidades cosechadas:", unidades)
````

## 2. `input()`

**¿Para qué sirve?**  
Solicita un dato al usuario por medio del teclado. El dato recibido se guarda inicialmente como texto.

**Sintaxis:**
```python
variable = input("Mensaje para el usuario: ")
```

**Ejemplo práctico:**  
Al ejecutar la celda, escribe Lechuga y presiona Enter. El resultado depende del texto que escriba el usuario. Si se necesita un número, después se puede convertir con int() o float().

````python
cultivo = input("Escribe el nombre del cultivo: ")
print("Cultivo registrado:", cultivo)
````

## 3. `type()`

**¿Para qué sirve?**  
Indica el tipo de dato de un valor o variable. Es útil para revisar si un dato es texto, entero, decimal u otro tipo.

**Sintaxis:**
```python
type(valor)
```

**Ejemplo práctico:**  
El resultado permite distinguir entre int (entero), float (decimal) y str (texto).

````python
unidades = 120
precio = 1250.5
cultivo = "Lechuga"
print(type(unidades))
print(type(precio))
print(type(cultivo))
````

## 4. `int()`

**¿Para qué sirve?**  
Convierte un valor compatible a un número entero. Cuando convierte un decimal, elimina la parte decimal; no redondea.

**Sintaxis:**
```python
int(valor)
```

**Ejemplo práctico:**  
La conversión es útil cuando un número llega como texto, por ejemplo desde input().

````python
unidades_texto = "120"
unidades = int(unidades_texto)
print(unidades)
print(type(unidades))
````

## 5. `float()`

**¿Para qué sirve?**  
Convierte un valor compatible a un número decimal.

**Sintaxis:**
```python
float(valor)
```

**Ejemplo práctico:**  
Se usa para trabajar con precios, pesos, medidas y cantidades que pueden tener decimales. En el código se utiliza punto decimal.

````python
precio_texto = "1250.50"
precio = float(precio_texto)
print(precio)
print(type(precio))
````

## 6. `str()`

**¿Para qué sirve?**  
Convierte un valor a una cadena de texto. Facilita unir números con mensajes.

**Sintaxis:**
```python
str(valor)
```

**Ejemplo práctico:**  
Sin str(), Python no permite concatenar directamente un número entero con una cadena usando el operador +.

````python
numero_lote = 27
mensaje = "El número del lote es " + str(numero_lote)
print(mensaje)
````

## 7. `bool()`

**¿Para qué sirve?**  
Convierte un valor a un dato lógico: True (verdadero) o False (falso).

**Sintaxis:**
```python
bool(valor)
```

**Ejemplo práctico:**  
En general, los textos no vacíos y los números diferentes de cero se consideran verdaderos. Los textos vacíos, cero y None se consideran falsos.

````python
nombre_cultivo = "Lechuga"
nombre_vacio = ""
print(bool(nombre_cultivo))
print(bool(nombre_vacio))
````

## 8. `len()`

**¿Para qué sirve?**  
Cuenta cuántos elementos hay en una colección o cuántos caracteres tiene un texto.

**Sintaxis:**
```python
len(objeto)
```

**Ejemplo práctico:**  
En el primer caso cuenta los elementos de la lista; en el segundo cuenta los caracteres del texto.

````python
cultivos = ["Lechuga", "Espinaca", "Acelga"]
print("Cantidad de cultivos:", len(cultivos))
print("Caracteres en Lechuga:", len("Lechuga"))
````

## 9. `range()`

**¿Para qué sirve?**  
Genera una secuencia de números enteros, muy utilizada para repetir acciones en ciclos for.

**Sintaxis:**
```python
range(inicio, fin, paso)
```

**Ejemplo práctico:**  
El valor final no se incluye. Por eso range(1, 6) produce 1, 2, 3, 4 y 5. Técnicamente range es un tipo de secuencia incorporado, no una función ordinaria.

````python
dias = list(range(1, 6))
print(dias)
````

## 10. `enumerate()`

**¿Para qué sirve?**  
Recorre una colección y entrega cada elemento junto con su posición o contador.

**Sintaxis:**
```python
enumerate(iterable, start=0)
```

**Ejemplo práctico:**  
El parámetro start=1 hace que la numeración comience en uno, algo útil para presentar una lista.

````python
cultivos = ["Lechuga", "Espinaca", "Acelga"]
for numero, cultivo in enumerate(cultivos, start=1):
    print(numero, cultivo)
````

## 11. `abs()`

**¿Para qué sirve?**  
Devuelve el valor absoluto de un número, es decir, su distancia respecto de cero sin signo negativo.

**Sintaxis:**
```python
abs(numero)
```

**Ejemplo práctico:**  
La diferencia se muestra como 7, aunque la resta original sea negativa o positiva.

````python
produccion_esperada = 120
produccion_real = 113
diferencia = abs(produccion_esperada - produccion_real)
print("Diferencia de unidades:", diferencia)
````

## 12. `round()`

**¿Para qué sirve?**  
Redondea un número a la cantidad de decimales indicada.

**Sintaxis:**
```python
round(numero, cantidad_de_decimales)
```

**Ejemplo práctico:**  
Con 2 como segundo argumento, el resultado se presenta con precisión de dos posiciones decimales. El redondeo de Python sigue la regla de empate hacia el número par.

````python
peso_caja = 12.678
print(round(peso_caja, 2))
````

## 13. `sum()`

**¿Para qué sirve?**  
Suma los elementos numéricos de una colección.

**Sintaxis:**
```python
sum(iterable, start=0)
```

**Ejemplo práctico:**  
Es útil para obtener totales a partir de listas de producción, ventas, costos o cantidades.

````python
cosecha_kg = [12, 15, 13, 10]
total_kg = sum(cosecha_kg)
print("Cosecha total:", total_kg, "kg")
````

## 14. `min()`

**¿Para qué sirve?**  
Devuelve el elemento de menor valor de una colección o el menor de varios argumentos.

**Sintaxis:**
```python
min(iterable)
```

**Ejemplo práctico:**  
Permite identificar el menor valor registrado en un conjunto de datos.

````python
temperaturas = [18.5, 20.0, 17.8, 21.2]
print("Temperatura mínima:", min(temperaturas), "°C")
````

## 15. `max()`

**¿Para qué sirve?**  
Devuelve el elemento de mayor valor de una colección o el mayor de varios argumentos.

**Sintaxis:**
```python
max(iterable)
```

**Ejemplo práctico:**  
Se puede utilizar para identificar el valor más alto de producción, temperatura, precio u otra variable.

````python
temperaturas = [18.5, 20.0, 17.8, 21.2]
print("Temperatura máxima:", max(temperaturas), "°C")
````
